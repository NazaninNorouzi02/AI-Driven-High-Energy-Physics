## **1.1 Positron Source**

For a typical PET scan, the injected activity is:  

$$A = 370\ \text{MBq} = 370 \times 10^6\ \text{decays/s} = 3.7 \times 10^8\ \text{s}^{-1}$$

Each positron annihilation produces two 511 keV gamma photons:

$$2A = 7.4 \times 10^8\ \gamma/\text{s}$$

However, only a small fraction of these gammas actually reach one detector module.  
If we assume about **1 % solid angle coverage (f = 0.01)**, then:

$$7.4 \times 10^8 \times 0.01 = 7.4 \times 10^6\ \gamma/\text{s}$$

So, a **reasonable representative incident flux** for one detector module is:

$$\boxed{1 \times 10^6\ \gamma/\text{s}}$$

## **1.5 Noise Discussion**

### **A) Sources of Noise**

1. **Photoelectron counting noise:**  
   Fundamental Poisson fluctuation in number of detected photoelectrons.

2. **PMT gain fluctuations:**  
   Dynode chain introduces extra variance (excess noise factor).

3. **Scintillator non-linearity:**  
   Light yield not perfectly proportional to energy deposition.

4. **Electronic/thermal noise:**  
   Johnson noise in resistors, amplifier input noise, etc.

5. **PMT dark current:**  
   Thermionic emission generating false photoelectrons.

6. **Pile-up and baseline fluctuations:**  
   Overlapping pulses and electronics drift during high count rates.

---

### **B) Comparison with Thermal Noise**

Johnson (thermal) RMS noise voltage:

$$V_{\text{rms}} = \sqrt{4 k_B T R \Delta f}$$

where  
$(k_B = 1.38 \times 10^{-23}\ \text{J/K}$),  
$(T = 300\ \text{K}$),  
$(R = 50\ \Omega$).

For bandwidth based on pulse duration:

$$\Delta f \approx \frac{1}{2\pi t}$$

**For LYSO (t = 40 ns):**

$$
\Delta f = \frac{1}{2\pi \times 40\times10^{-9}} \approx 3.98 \times 10^6\ \text{Hz}
$$

Plugging in:

$$
V_{\text{rms}} = \sqrt{4 \times 1.38 \times 10^{-23} \times 300 \times 50 \times 3.98\times10^6} \approx 1.8\times10^{-6}\ \text{V} = 1.8\ \mu\text{V}
$$

**For EJ-301 (t = 10 ns):**

Bandwidth ≈ 16 MHz → noise ≈ 2× higher → about **3.6 µV**.

---

Compare to signal and statistical noise:

- EJ-301 signal ≈ 150 mV amplitude  
- Counting fractional σ ≈ 3.7 % → σ ≈ 5.6 mV  
- Johnson noise ≈ 1–3 µV → **over 1000× smaller**

Statistical fluctuations and PMT gain noise dominate.  
Thermal/electronic noise is negligible in this PET detection scenario.

### Exercise 02: CCD Noise Estimation with Finite Background Sampling  

###### A.
####What is CCD?

A Charge-Coupled Device (CCD) is an array of light-sensitive semiconductor pixels.  
Each pixel acts as a small potential well that **collects photoelectrons** produced when incident photons are absorbed in the silicon substrate.  
The number of collected electrons is proportional to the number of photons that arrive and to the **quantum efficiency** (η_d).  
After the exposure, the accumulated charge in each pixel is shifted through the device by means of clocked gate voltages, the **charge-coupling process**, and finally converted to a voltage by an on-chip amplifier and digitized.  

Main parameters used here:  

| Symbol | Meaning |
|:--:|:--|
| η_d | responsive quantum efficiency (fraction of photons detected) |
| $N_{νS}$ | total number of incident signal photons (summed over all signal pixels) |
| $N_{νB}$ | mean number of background photons per pixel during the exposure |
| $n_{pix}$ | number of pixels integrated in the photometric aperture |



The CCD is assumed *ideal*: no read noise, no dark current, and perfect charge-transfer efficiency.

---
#### Assumptions used  

1. Photon arrivals obey Poisson statistics (shot-noise limited).  
2. Quantum efficiency $η_{d}$ is constant and identical for signal and background photons.  
3. Read noise, dark current, and fixed-pattern noise are neglected (ideal CCD).  
4. Background is estimated from many pixels so its sampling uncertainty is negligible.  

---

#### Signal and noise model  

During an exposure each pixel receives both signal and background photons.  
Photon arrival follows **Poisson statistics**, therefore the variance equals the mean number of detected electrons.  

- Detected **signal electrons**:  
  $$ S = \eta_d N_{νS} $$

- Detected **background electrons** per pixel:  
  $$ B = \eta_d N_{νB} $$

If the number of pixels are summed in the aperture, the total detected counts are :
$$
C_{\text{total}} = \eta_d N_{νS} + \eta_d n_{\text{pix}} N_{νB}
$$

The **net source signal** after background subtraction is :
$$
S_{\text{net}} = \eta_d N_{νS}
$$

---

####  Noise variance  

For an ideal CCD the dominant noise sources are the Poisson fluctuations of the detected photons:

$$
\sigma^2_{\text{total}}
 = \eta_d N_{νS}           \quad \text{(signal shot noise)}
 + \eta_d n_{\text{pix}} N_{νB}  \quad \text{(background shot noise)}.
$$

Any uncertainty in the background estimation due to finite sampling is small if the background is measured from a large surrounding area, so it is neglected here.

---

####   Signal-to-Noise Ratio  

The definition :
$$
\mathrm{SNR} = \frac{\text{net signal}}{\text{rms noise}}
$$
gives :
$$
\mathrm{SNR}
  = \frac{\eta_d N_{νS}}
         {\sqrt{\eta_d N_{νS} + \eta_d n_{\text{pix}} N_{νB}}}
  = \frac{\sqrt{\eta_d}\,N_{νS}}
         {\sqrt{N_{νS} + n_{\text{pix}} N_{νB}}}.
$$

Factorising $N_{νS}$ inside the denominator:

$$
\boxed{
\mathrm{SNR}
 = \frac{\sqrt{\eta_d\,N_{νS}}}
        {\sqrt{1 + n_{\text{pix}}\left(\dfrac{N_{νB}}{N_{νS}}\right)}}
}.
$$

---

####  Interpretation  

- For **signal-dominated** exposures :  
  $$ n_{\text{pix}} N_{νB} \ll N_{νS} $$
  $$\mathrm{SNR} \approx \sqrt{\eta_d N_{νS}}$$

- For **background-dominated** cases :
  $$ n_{\text{pix}} N_{νB} \gg N_{νS} $$  
  $$\mathrm{SNR} \approx \dfrac{\eta_d N_{νS}}{\sqrt{\eta_d n_{\text{pix}} N_{νB}}}$$

Thus, the signal-to-noise ratio grows as the square root of detected signal photons but decreases with increasing background level or aperture size.



#### B.
#### **Background estimated from a finite number of background pixels:**

In Part A we assumed that the background level was measured from a very large number of background pixels near the source.  
That means the average background value was known very precisely, so the uncertainty in this estimate could be ignored.  

Now we consider a more realistic case where the background is estimated from a finite number of background pixels, noted as $n_{bck}$, which may be comparable to the number of signal pixels $n_{pix}$.

---

Each pixel in the aperture receives both signal and background photons, and both follow Poisson statistics.  
After background subtraction, the total uncertainty (variance) of the net signal comes from three parts:

1. **Shot noise of the signal photons:**  
   $$\sigma^2_S = \eta_d N_{\nu S}$$

2. **Shot noise of the background photons inside the source aperture:**  
   $$\sigma^2_B = \eta_d n_{\text{pix}} N_{\nu B}$$

3. **Uncertainty of the estimated mean background (finite sample):**  
   The mean background per pixel is measured from $n_{\text{bck}}$ pixels,  
   so its variance is $\dfrac{\eta_d N_{\nu B}}{n_{\text{bck}}}$,  
   and when this mean is subtracted from $n_{\text{pix}}$ source pixels the total variance becomes  
   $$\sigma^2_{\text{bck,est}} = n_{\text{pix}}^2 \dfrac{\eta_d N_{\nu B}}{n_{\text{bck}}}$$

---

Adding these three contributions gives the total noise variance:

$$
\sigma^2_{\text{total}} =
\eta_d N_{\nu S}
+ \eta_d n_{\text{pix}} N_{\nu B}
+ n_{\text{pix}}^2 \dfrac{\eta_d N_{\nu B}}{n_{\text{bck}}}
$$

The signal-to-noise ratio is the net detected source divided by the rms noise:

$$
\mathrm{SNR} =
\dfrac{\eta_d N_{\nu S}}
{\sqrt{\eta_d N_{\nu S}
+ \eta_d n_{\text{pix}} N_{\nu B}
+ n_{\text{pix}}^2 \dfrac{\eta_d N_{\nu B}}{n_{\text{bck}}}}}
$$

Factoring $\eta_d N_{\nu S}$ gives a more compact form:

$$
\mathrm{SNR} =
\dfrac{\sqrt{\eta_d N_{\nu S}}}
{\sqrt{\,1
+ n_{\text{pix}}\dfrac{N_{\nu B}}{N_{\nu S}}
+ \dfrac{n_{\text{pix}}^2}{n_{\text{bck}}}\dfrac{N_{\nu B}}{N_{\nu S}}\,}}
$$

If $n_{\text{bck}}$ is very large, the last term disappears and this reduces to the result from Part A.

---

**Simple explanation — how A and B differ:**

In Part A the background was measured from many pixels, so the uncertainty in the average background was tiny and could be ignored.  
In Part B we measure the background from only a few pixels, so the mean background value we subtract is noisy.  
That uncertainty in the estimated background adds another noise term to the denominator, which reduces the SNR.  
The smaller the number of background pixels, the greater this effect.

---

#### C.

**Including dark current and read noise:**

In a real CCD there are two additional sources of noise besides photon shot noise:

1. **Dark current** which are electrons thermally generated inside each pixel even with no light.  
   Let $I$ be the dark current per pixel (electrons per unit time).  
   For exposure time $t$, the mean dark charge per pixel is $I t$ with variance $I t$.

2. **Read noise** which is random uncertainty in the charge measurement introduced by the on-chip amplifier and electronics.  
   Denote its RMS per pixel by $\Delta q_R$.  
   It is independent of exposure time.

Including these effects, the total noise variance becomes:

$$
\sigma^2_{\text{total}} =
\eta_d N_{\nu S}
+ \eta_d n_{\text{pix}} N_{\nu B}
+ n_{\text{pix}}^2 \dfrac{\eta_d N_{\nu B}}{n_{\text{bck}}}
+ n_{\text{pix}} I t
+ n_{\text{pix}} (\Delta q_R)^2
$$

and the SNR is:

$$
\mathrm{SNR} =
\dfrac{\eta_d N_{\nu S}}
{\sqrt{\eta_d N_{\nu S}
+ \eta_d n_{\text{pix}} N_{\nu B}
+ n_{\text{pix}}^2 \dfrac{\eta_d N_{\nu B}}{n_{\text{bck}}}
+ n_{\text{pix}} I t
+ n_{\text{pix}} (\Delta q_R)^2}}
$$

---

**Dependence on integration time:**

If the signal grows linearly with time, $N_{\nu S} = s t$ (where $s$ is the photon rate):

- The **read noise term** scales as  
  $$\dfrac{n_{\text{pix}}(\Delta q_R)^2}{\eta_d N_{\nu S}}
  = \dfrac{n_{\text{pix}}(\Delta q_R)^2}{\eta_d s t}
  \propto \dfrac{1}{t}$$  
  so the effect of read noise decreases with longer exposure.

- The **dark current term** scales as  
  $$\dfrac{n_{\text{pix}} I t}{\eta_d N_{\nu S}}
  = \dfrac{n_{\text{pix}} I}{\eta_d s}$$  
  which is independent of $t$.  
  This means the dark current contribution to noise does **not** become smaller with longer exposure; it grows at the same rate as the signal itself.

---

**What I learned from this exercise:**

- Increasing the number of background pixels reduces the uncertainty in the background subtraction.  
- Read noise becomes less important for longer exposures because it does not scale with time.  
- Dark current increases linearly with time, so its contribution stays constant relative to the signal.  
- In practice, to reach the shot-noise limit, one wants long enough exposures that read noise is negligible but not so long that dark current dominates.

# Exercise 04: Maximum Likelihood Estimator

Consider n data points draw independently from a gaussian distribution. Starting from maximum likelihood estimator, find that it equivalent to minimizing the sum-of-squares error function.


#### Your answer:

---
### Exercise 04: Maximum Likelihood Estimator (MLE) and Sum of Squares

We have $n$ independent data points $x_1, x_2, \dots, x_n$ drawn from a Gaussian (normal) distribution:

$$
x_i \sim \mathcal{N}(\mu, \sigma^2)
$$

where $\mu$ and $\sigma^2$ are the unknown mean and variance.

The probability density function is:

$$
p(x_i | \mu, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left[-\frac{(x_i - \mu)^2}{2\sigma^2}\right]
$$

---

### Step 1: Likelihood function

Since the data points are independent, the total likelihood is the product of all individual probabilities:

$$
L(\mu, \sigma^2) = \prod_{i=1}^n p(x_i | \mu, \sigma^2)
= \prod_{i=1}^n \frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left[-\frac{(x_i - \mu)^2}{2\sigma^2}\right]
$$

---

### Step 2: Log-likelihood

To simplify, take the natural logarithm:

$$
\ell(\mu, \sigma^2) = \log L(\mu, \sigma^2)
= \sum_{i=1}^n \left[-\frac{1}{2}\log(2\pi\sigma^2)
- \frac{(x_i - \mu)^2}{2\sigma^2}\right]
$$

Simplify the expression:

$$
\ell(\mu, \sigma^2) = -\frac{n}{2}\log(2\pi\sigma^2)
- \frac{1}{2\sigma^2}\sum_{i=1}^n (x_i - \mu)^2
$$

---

### Step 3: Find MLE for $\mu$

To find $\hat{\mu}_{\text{MLE}}$, we fix $\sigma^2$ (treat it as constant) and maximize $\ell$ with respect to $\mu$.

Take derivative w.r.t. $\mu$:

$$
\frac{\partial \ell}{\partial \mu}
= -\frac{1}{2\sigma^2} \cdot 2\sum_{i=1}^n (\mu - x_i)
= -\frac{1}{\sigma^2}\sum_{i=1}^n (\mu - x_i)
$$

Set derivative equal to zero:

$$
\sum_{i=1}^n (\mu - x_i) = 0
\quad \Rightarrow \quad
n\mu = \sum_{i=1}^n x_i
$$

So:

$$
\hat{\mu}_{\text{MLE}} = \frac{1}{n}\sum_{i=1}^n x_i
$$

This is just the **sample mean**.

---

### Step 4: Connection to the sum-of-squares error

The log-likelihood can be written as:

$$
\ell(\mu) = \text{constant} - \frac{1}{2\sigma^2}\sum_{i=1}^n (x_i - \mu)^2
$$

Maximizing $\ell(\mu)$ is equivalent to **minimizing** the second term:

$$
E(\mu) = \sum_{i=1}^n (x_i - \mu)^2
$$

This $E(\mu)$ is exactly the **sum-of-squares error** function.

Therefore, the **MLE estimate of $\mu$** is the value that **minimizes the sum of squared differences** between the data and the mean.
